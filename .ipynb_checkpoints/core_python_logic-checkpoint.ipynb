```python
pdc = 76
if pdc >= 80:
    print("Patient is adherent!")
else: 
    print("Patient is not adherent!")
```

---
The code calculates Proportion of Days Covered (PDC), which is a measure of medication adherence. A patient must have a PDC value of 80 or greater to be adherent.
Otherwise, the patient is not adherent to medication.

```python
patients = 0
for patient in range(5):
    patients += 1
    print("Patient:", patient)
    print("Cumulative patients:", patients)
```

---
The code is counting 5 patients, adding a patient to cumulative patients each time the loop runs.
